```python
candidatos = {
    1: "jojo todinho",
    2: "fulano",
    3: "ciclano"
}

# Quantidade de votos de cada candidato
votos = {
    1: 0,
    2: 0,
    3: 0
}

# Contadores
total_eleitores = 0
total_votaram = 0


# Início do cadastro dos eleitores
continuar = "s"

while continuar == "s":

    print("\n===== CADASTRO DO ELEITOR =====")

    nome = input("Digite o nome do eleitor: ")
    idade = int(input("Digite a idade do eleitor: "))

    total_eleitores += 1

    # Verifica se o eleitor pode votar
    if idade < 18:

        print("\n", nome, "não pode votar.")
        print("É necessário ter 18 anos ou mais.")

    else:

        print("\n", nome, "pode votar!")

        print("\n===== CANDIDATOS =====")

        for numero, nome_candidato in candidatos.items():
            print(numero, "-", nome_candidato)

        # Permite tentar novamente caso o voto seja inválido
        while True:

            voto = int(input("\nDigite o número do candidato: "))

            if voto in candidatos:

                # Adiciona um voto ao candidato escolhido
                votos[voto] += 1

                # Aumenta a quantidade de pessoas que votaram
                total_votaram += 1

                print("\nVoto registrado com sucesso!")
                break

            else:

                print("\nCandidato inválido!")
                print("Digite novamente um número válido.")


    # Pergunta se deseja cadastrar outro eleitor
    continuar = input("\nDeseja cadastrar outro eleitor? (s/n): ").lower()


# ==============================
# RESULTADOS
# ==============================

print("\n================================")
print("         RESULTADO DA ELEIÇÃO")
print("================================")

print("\nTotal de eleitores cadastrados:", total_eleitores)
print("Total de eleitores que votaram:", total_votaram)

print("\nVotos de cada candidato:")

for numero, nome_candidato in candidatos.items():

    print(nome_candidato, "-", votos[numero], "votos")


# Descobrindo a maior quantidade de votos
maior_voto = max(votos.values())


# Criando uma lista para guardar os vencedores
vencedores = []

for numero, quantidade in votos.items():

    if quantidade == maior_voto:

        vencedores.append(candidatos[numero])


# Verificando se houve empate
if len(vencedores) == 1:

    print("\n================================")
    print("VENCEDOR:", vencedores[0])
    print("Quantidade de votos:", maior_voto)
    print("================================")

else:

    print("\n================================")
    print("HOUVE EMPATE!")
    print("================================")

    print("Candidatos empatados:")

    for vencedor in vencedores:
        print("-", vencedor)

    print("Quantidade de votos:", maior_voto)
```
